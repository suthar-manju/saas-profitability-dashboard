```python
import pandas as pd, numpy as np
from scipy import stats

df = pd.read_csv(r"C:\Saas-Sales-Profitability\Data\Clean\saas_clean.csv")
print(df.shape)
print("Total profit:", round(df["Profit"].sum()))
print("Profit above 20% discount:", round(df.loc[df["Discount"] > 0.2, "Profit"].sum()))
```

```python
r, p = stats.pearsonr(df["Discount"], df["Profit"])
rho, p2 = stats.spearmanr(df["Discount"], df["Profit"])
print("Pearson r =", round(r, 3), " p =", p)
print("Spearman rho =", round(rho, 3), " p =", p2)
```

```python
groups = [g["Profit"].values for _, g in df.groupby("Discount_Tier")]
H, p = stats.kruskal(*groups)
print("Kruskal-Wallis H =", round(H, 1), " p =", p)
```

```python
res = stats.linregress(df["Discount"], df["Profit"])
print("Change in profit per line for each 10-point rise in discount:", round(res.slope * 0.1, 1))
print("R-squared:", round(res.rvalue ** 2, 3))
```

```python
from sklearn.tree import DecisionTreeClassifier

tree = DecisionTreeClassifier(max_depth=1, random_state=42)
tree.fit(df[["Discount"]], df["Is_Loss"])
print("Split at discount =", round(tree.tree_.threshold[0] * 100, 1), "%")
```

```python
cap = 0.20
over = df["Discount"] > cap
cost = df["Sales"] - df["Profit"]
list_price = df["Sales"] / (1 - df["Discount"])
new_sales = np.where(over, list_price * (1 - cap), df["Sales"])
new_profit = (new_sales - cost).sum()

print("Lines affected:", int(over.sum()))
print("Current profit:", round(df["Profit"].sum()))
print("Profit with cap:", round(new_profit))
print("Difference:", round(new_profit - df["Profit"].sum()))
```
